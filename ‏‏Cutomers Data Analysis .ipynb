```python
df['Review Rating']=df.groupby(df['Category'])['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
```

```python
df=df.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
```

```python
df.columns
```

```python
# create columns age_group
labels=['young adult', 'adult', 'middle_aged', 'senior']
df['age_group']=pd.qcut(df['age'] , q=4, labels=labels)
```

```python
df[['age', 'age_group']].sample(10)
```

```python
# create column purchase_frequency_days 
df['frequency_of_purchases'].value_counts()
```

```python
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Bi-Weekly':14,
    'Quarterly':90,
    'Every 3 Months':90,
    'Annually':365
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
    
```
